### **Tarea A.2:** Visualizar la imagen con ``cv2.imshow()``


```python
# TODO: Display the image using cv2.imshow().
cv2.imshow("First image", img)  # a imshow() tengo que pasarle los argumentos del nombre que quiero darle a la imagen, y la propia imagen (Mat, de matriz)
cv2.waitKey(0)  # en waitKey() ponemos la tecla del teclado con la que queremos cerrar la imagen
cv2.destroyAllWindows() # esto cierra todas las ventanas de despliegue de imágenes que hayamos abierto antes
```

### **Tarea A.3:** Conversion de imagen de color a escala de grises ``cv2.cvtColor()``.

```python
# TODO: Convert the image from color to grayscale using cv2.cvtColor().
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)  # cvtColor toma la imagen y el código de color al que quieres convertir la imagen (en este caso a grises)

"""
Para pasar de RGB a escala de grises (un solo canal), cv2 hace por dentro una media ponderada 0.6 * R + 0.2 * G + 0.1 * B

En open cv es BGR en vez de RGB, cuidado con eso que es muy importante para determinadas tareas.

"""

# TODO: Check image shape → grayscale: (height, width), color: (height, width, channels).
print("Color shape:", img.shape)  # el atributo shape muestra el alto, ancho y el número de canales de color de los que dispongo
print("Gray shape:", gray.shape)  # en la shape de la imagen en gris solo mostrará 2 dimensiones: la altura y el ancho, ya que el canal ahora ya es solo 1 en vez de 3
# TODO: Display the image using cv2.imshow().
cv2.imshow("Gray image", gray )
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### **Tarea A.4:** Cambia el tamaño de una imagen ``cv2.resize()`` a 200x200 píxeles.

```python
# TODO: Resize the image to 200×200 pixels using cv2.resize().
new_height_size, new_width_size = 200, 200
resized = cv2.resize(src=img, dsize=(new_height_size, new_width_size) )  # le tengo que pasar la imagn (matriz) y en una tupla, el tamaño al que lo quiero redimensionar
# TODO: Check the image shape
print("Resized shape:", resized.shape)  # mostrará que ahora las dimensiones de altura y anchura, tienen un rango distinto porque lo hemos redimensionado
# TODO: Display the image using cv2.imshow().
cv2.imshow("Resized image", resized)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### **Tarea A.5:** Recorta el centro de la imagen a un tamaño 120x160.

```python
# TODO: # TODO: Crop the center of the image 'img' with an approximate size of 120x160 pixels.
h, w = img.shape[:2]  # calculamos cuánto vale el total de la altura y anchura de la imagen
cy, cx = h//2, w//2  # tomamos la mitad de la altura y la mitad de la anchura, y esa tupla de puntos nos define el centro de la imagen
new_h, new_w = 120, 160  # definimos cuánto queremos que mida la imagen recortada, el ROI
half_of_new_h, half_of_new_w = new_h//2, new_w//2  # calculamos cuánto tenemos que sumar y restar desde el centro para crear nuestro recorte

# Definimos los límites de las nuevas filas y columnas de la matriz recortada
new_top_row, new_down_row = cy-half_of_new_h, cy+half_of_new_h
new_left_col, new_right_col = cx-half_of_new_w, cx+half_of_new_w

# Recortamos la imagen
roi = img[new_top_row:new_down_row, new_left_col:new_right_col]

"""

Como las imágenes son matrices (arrays), con filas y columnas, si queremos mostrar solo algunos píxeles determinados de la imagen, o recortarla como en este caso, lo que
tenemos que hacer es hacer slicing seleccionando qué filas y qué columnas queremos usar.

Si usamos "new_top_row:new_down_row" estamos seleccionando desde la fila de "más arriba" en la que queremos que sea la nueva matriz, 
hasta la fila "más abajo" también de la nueva matriz. Por eso usamos los ":".

Para las columnas, hacemos lo mismo.

"""

# TODO: Check the image shape
print("ROI shape:", roi.shape)  # el ROI es la "Region of Interest" (comprobación de que la croped image mide lo deseado)
# TODO: Display the image using cv2.imshow().
cv2.imshow("Croped image", roi)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### **Tarea A.6:** Rota la imagen usando ``cv2.getRotationMatrix2D`` y ``cv2.wrapAffine()``.

```python
# TODO: Rotate around the center using cv2.getRotationMatrix2D and cv2.warpAffine.

M = cv2.getRotationMatrix2D(center=(cy,cx), angle=90, scale=1)  # los ángulos hay que pasarlos en grados (positivos significa sentido antihorario)
# la función de getRotationMatrix2D, calcula la matriz de rotación para una imagen en 2 D

print(M) 

rotated = cv2.warpAffine(src=img, M=M, dsize=img.shape[:2])
# la función warpAffine aplica una transformación afín a una imagen (en este caso un giro)

# TODO: Display the image using cv2.imshow().
cv2.imshow("Rotated image", rotated)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### **Tarea A.7:** Guarda la imagen con ``cv2.imwrite()``

```python
# TODO: Save the image to disk as 'salida_opcionA_1.png' using cv2.imwrite().
save_img_path = "salida_opcionA_1.png"
cv2.imwrite(save_img_path, img)  # a la función de imwrite hay que pasarle el path donde queremos guardar la imagen, y la propia matriz de la imagen
```

### **Tarea B.2**. Defina los métodos ``show_image()`` y ``write_image()`` para visualizar y guardar imágenes

```python
# TODO Complete the method, use every argument
def show_image(img: np.array, img_name: str = "Image"):
    cv2.imshow(img_name, img)  # mostramos la imagen, poniendole el nombre correspondiente a la ventana desplegada
    cv2.waitKey(0)  # ponemos una tecla de introducción
    cv2.destroyAllWindows()

# TODO Complete the method, use every argument
def write_image(output_folder: str, img_name: str, img: np.array):
    img_path = os.path.join(output_folder, img_name)  # creamos el path donde vamos a guardar la imagen
    os.makedirs(output_folder, exist_ok=True)
    cv2.imwrite(img_path, img)  # guardamos la imagen en su correspondiente ubicación
```

### **Tarea B.4:** Segmente los colores anaranjados

```python
# TODO Define orange color range
light_orange = (0, 200, 120)  # son los del min
dark_orange = (255, 255, 255)   # son los del max

# TODO Compute a list that contains a mask (which should segment orange colors) for every image.
orange_masks = [cv2.inRange(hsv_im, light_orange, dark_orange) for hsv_im in hsv_imgs]

"""

La función de inRange lo que hace es comprobar si cada elemento del primer array que le pasas, cumple que sus valores están en el rango
que delimitan los otros 2 arrays que le pasas. Hace la verificación para los 3 canales (HSV).

Devuelve 255 (True) o 0 (False), en función de si el pixel está en el rango o no, y así se crea la máscara booleana. 
La estructura que devuelve cada vez, es de la misma altura y anchura que la imagen que se le había pasado (matriz)

"""

# TODO Compute a list that contains the result of multiplying the original image with its orange colors mask.
orange_segmented = [cv2.bitwise_and(img, img, mask=mask)
                        for img, mask in zip(imgs, orange_masks)]

"""

La función de bitwise_and lo que hace es hacer la conjunción/intersección pixel a pixel, del pixel de img con el pixel de img,
es decir de un mismo pixel consigo mismo, pero lo hace solo en el caso donde la máscara en ese bit sea True.

Entonces como cada pixel se representa por una serie de bits, la intersección de un pixel consigo mismo da siempre el mismo pixel (los mismos bits) pero ese
resultado solo se pone cuando la mask (booleana) deja (cuando la mask está en 255, True, y sino es porque está en False, 0)

"""

n = 5

# TODO Show an original image
show_image(imgs[n], "Original image")

# TODO Show a mask
show_image(orange_masks[n], "Orange Masks")

# TODO Show a segmented image
show_image(orange_segmented[n], "Orange segmented image")
```

### **Tarea B.5:** Segmente los colores blancos.

Para detectar el rango de blancos complete la siguiente celda y ejecutela para investigar el rango de valores necesarios.

```python
# TODO Discover white color ranges
# hsv_ranges = [get_hsv_color_ranges(im) for im in imgs]
```

```python
# TODO Define white color range
light_white = (0, 0, 180)
dark_white = (170, 175, 255)

# TODO Compute a list that contains a mask (which should segment white colors) for every image.
white_masks = [cv2.inRange(hsv_im, light_white, dark_white) for hsv_im in hsv_imgs]

# TODO Compute a list that contains the result of multiplying the original image with its white colors mask.
white_segmented = [cv2.bitwise_and(img, img, mask=mask)
                        for img, mask in zip(imgs, white_masks)]
n = 4

# TODO Show an original image
show_image(imgs[n], "Original image")

# TODO Show a mask
show_image(white_masks[n], "White Mask")

# TODO Show a segmented image
show_image(white_segmented[n], "White segmented image")
```

### **Tarea B.6:** Junte las máscaras para cada imagen (naranja + blanco) y segmente cada una de ellas.

```python
# TODO Join orange_masks and white_masks
fish_masks = [cv2.bitwise_or(o,w) for o, w in zip(orange_masks, white_masks)]

# el bit wise or, es una puerta lógica, que dice que si en alguna de las dos imagenes el pixel es naranja o blanco, da true 
# Eso es porque comprueba las máscaras booleanas de ambos colores, y con que una dé True, ya da True el OR
    
# TODO Compute a list that contains the result of multiplying the original image with its complete mask.
fish = [cv2.bitwise_and(img, img, mask=mask) for img, mask in zip(imgs, fish_masks)]
# Aquí cogemos los pixeles donde la máscara de naranja o blanco haya dado True

n = 4

# TODO Show an original image
show_image(imgs[n], "Original image")

# TODO Show a mask
show_image(fish_masks[n], "Fish Mask")

# TODO Show a segmented image
show_image(fish[n], "White & Orange segmented image")
```

### **Tarea B.7:** Guarde las imágenes

```python
# TODO Define your output folder and save every fish image
output_folder = "output_folder"
for i in range((len(fish))):
    img_name = f"fish_{i}.png"
    write_image(output_folder=output_folder, img_name=img_name, img=fish[i])
```

### **Pregunta B.2:** ¿Qué ocurre si carga las imágenes con la función ``imageio.read()`` y luego la muestra con el método ``show_image()``? ¿A qué se debe este comportamiento?

```python
def load_images(filenames: List) -> List:
    return [imageio.imread(filename) for filename in filenames]

imgs_path = ["../data/*.jpg"] # con esto podemos cargar todas las imágenes de la carpeta data que acaben en jpg
imgs_path = glob.glob(imgs_path[0]) # el glob se usa para bu8scar la ruta 

# TODO Homework: Load images
question_imgs = load_images(imgs_path)

# TODO Homework: Show it
for i, im in enumerate(question_imgs):
    show_image(im, f"Question image {i}")
```

### Respuesta:

Al cargar una imagen con **imageio.imread()**, los valores de los píxeles se almacenan en formato RGB. 

Sin embargo, cuando la mostramos con show_image() (que internamente utiliza OpenCV en **cv2.imshow()**), la librería interpreta las imágenes en formato BGR.

Esta diferencia en el orden de los canales provoca que las imágenes aparezcan con colores distintos a los que veíamos cuando cargábamos las imágenes con **cv2.imread()** porque ahora donde teníamos R, tenemos B, es decir se han intercambiado los canales rojo y azul. Es por ello, que ahora vemos las imágenes con tintes azulados, y lo que antes eran peces anaranjados y blancos, ahora son azulados y blancos.

Para corregir esto en caso de que quisiéramos seguir usando **imageio.imread()**, podríamos convertir explícitamente el espacio de color de RGB a BGR antes de mostrar la imagen, como se muuestra en el siguiente código:



```python
img_bgr = cv2.cvtColor(question_imgs[0], cv2.COLOR_RGB2BGR)  # convertimos de RGB a BGR, para que OpenCV ya no genere más problemas
show_image(img_bgr, "Example image")
```
